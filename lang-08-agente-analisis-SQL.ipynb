```python
# Creación de Agente para anaísis automático SQL a partir de consultas en lenguaje natura
# Ejemplo con BBDD MySQL
# https://docs.langchain.com/oss/python/langchain/sql-agent
import os
from langchain_mistralai.chat_models import ChatMistralAI
from langchain_core.prompts import ChatPromptTemplate, PromptTemplate, SystemMessagePromptTemplate, HumanMessagePromptTemplate
from langchain_classic.agents import load_tools,initialize_agent,AgentType,create_react_agent,AgentExecutor

os.environ["MISTRAL_API_KEY"] = open('mistral_key_01.txt').read()

model_name = "mistral-large-latest"
temperature = 0
retries = 2

llm = ChatMistralAI(model=model_name, temperature=temperature, max_retries=retries)
```

```python
#
# pip install mysql-connector-python
#
import mysql.connector
```

```python
f = open('password_sql.txt')
password_sql = f.read()

config = {
    'user': 'usuario',
    'password': password_sql,
    'host': 'mysql-host',
    'database': 'world'
}
```

```python
conn = mysql.connector.connect(**config)
cursor = conn.cursor()
```

```python
# Ejecución de la consulta manualmente
query = """SELECT SUM(Population) FROM Country WHERE Continent = 'Asia';"""

cursor.execute(query)
result = cursor.fetchone()
```

```python
suma_poblacion = result[0] if result[0] is not None else 0
printf(f"La suma de la población del continente Asia es: {suma_poblacion}")
```

```python
# Crear el Agente SQL
# https://docs.langchain.com/oss/python/langchain/sql-agent
#
from langchain_community.agent_toolkits import create_sql_agent
from langchain_community.utilities import SQLDatabase
```

```python
connection_string = f"mysql+mysqlconnector://{config['user']}:{config['password']}@{config['host']}//{config['database']}"

db = SQLDatabase.from_uri(connection_string)
```

```python
agente = create_sql_agent(
    llm,
    db=db,
    verbose=True
)
```

```python
agente.invoke("Dime la población total de Asia")
```

```python
agente.invoke("Dime el promedio de la esperanza de vida por cada una de las regiones ordenadas de mayor a menor")
```
